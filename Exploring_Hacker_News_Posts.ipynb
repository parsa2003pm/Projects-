**Exploring Hacker News Posts**

Hacker News is a popular technology news site where users submit links and ask questions. You’ll analyze two post types, “Ask HN” and “Show HN” to figure out what makes a post get traction. Clean data, real questions, solid practice with Python string handling and datetime analysis.

*Step-by-Step Instructions:*

1.  Load the CSV and explore the columns:
title, URL, points, comments, time.
2.  Separate posts into “Ask HN”, “Show HN”, and other categories.
3.  Calculate average comments for each type.
4.  Parse the created_at column using the datetime module.
5.  Find which hours produce the most comments for Ask HN posts.
6.  Summarize findings in plain language.

```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("hacker-news/hacker-news-posts")

print("Path to dataset files:", path)
```

1.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('HN_posts_year_to_Sep_26_2016.csv')
df.head(5)
```

```python
df = df[['title', 'url', 'num_points', 'num_comments', 'created_at']]
df.rename(columns={'num_points': 'points', 'url' : 'URL', 'num_comments': 'Comments', 'created_at': 'Time'}, inplace=True)
df.head(5)
```

2.

```python
def categorize_title(title):
    if str(title).lower().startswith('ask hn'):
        return 'ask hn'
    elif str(title).lower().startswith('show hn'):
        return 'show hn'
    else:
        return 'others'

df['title_category'] = df['title'].apply(categorize_title)
print(df['title_category'].value_counts())
df.head(5)
```

3.

```python
a = df[df['title_category'] == 'ask hn']['Comments'].mean()
b = df[df['title_category'] == 'show hn']['Comments'].mean()
c = df[df['title_category'] == 'others']['Comments'].mean()
print("Ask HN average comments:", a)
print("Show HN average comments:", b)
print("Others average comments:", c)
```
