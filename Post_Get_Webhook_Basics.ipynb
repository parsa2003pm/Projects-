```python
import requests
```

```python
url_CityInfos = 'https://geocoding-api.open-meteo.com/v1/search'
response = requests.get(url_CityInfos)
print(response.status_code)
```

```python
name = input('enter your city name: ')
```

```python
params_Infos = {'name': name}
response = requests.get(url_CityInfos, params=params_Infos)
data = response.json()
data = data['results'][0]
latitude = data['latitude']
longitude = data['longitude']
print(latitude, longitude)
```

```python
url_CityWeather = 'https://api.open-meteo.com/v1/forecast'
response = requests.get(url_CityWeather)
print(response.status_code)
```

```python
params_weather = {'latitude': latitude, 'longitude': longitude, 'current_weather': True}
response = requests.get(url_CityWeather, params=params_weather)
data = response.json()
weather_codes = {
    0: "Clear sky",
    1: "Mainly clear",
    2: "Partly cloudy",
    3: "Overcast",
    45: "Fog",
    48: "Depositing rime fog",
    51: "Light drizzle",
    53: "Moderate drizzle",
    55: "Dense drizzle",
    56: "Light freezing drizzle",
    57: "Dense freezing drizzle",
    61: "Slight rain",
    63: "Moderate rain",
    65: "Heavy rain",
    66: "Light freezing rain",
    67: "Heavy freezing rain",
    71: "Slight snowfall",
    73: "Moderate snowfall",
    75: "Heavy snowfall",
    77: "Snow grains",
    80: "Slight rain showers",
    81: "Moderate rain showers",
    82: "Violent rain showers",
    85: "Slight snow showers",
    86: "Heavy snow showers",
    95: "Thunderstorm",
    96: "Thunderstorm with slight hail",
    97: "Heavy thunderstorm",
    99: "Thunderstorm with heavy hail",
}
a = data['current_weather']['temperature']
b = weather_codes[data['current_weather']['weathercode']]
print(f'the temperature in {name} is {a} and the weather is {b}')
```

```python
url_post = 'https://postman-echo.com/post'
weather_jason = {
    'temperature': a,
    'weather': b
}
response = requests.post(url_post, json=weather_jason)
print(response.status_code)
```

```python
API_KEY = 'test_key_12345'
headers = {
    'Authorization': f'Bearer {API_KEY}'
}
response = requests.post(url_post, json=weather_jason, headers=headers)
print(response.status_code)
result = response.json()
print('Sent data echoed back:', result['json'])
```

```python
assert result['json'] == weather_jason, 'data mismatch'
assert result['headers']['authorization'] == f'Bearer {API_KEY}', 'header mismatch'
print('OK: data and header both match')
```
